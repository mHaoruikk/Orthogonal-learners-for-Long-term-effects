# Synthetic PEHE results across overlap regimes

All runs use the old DGP (`use_old_dgp: true`), Y normalization, and the
shared learner roster. B=5 replicates per dataset, master_seed=42, NN
(torch_mlp) backbone for the CATE regressor.

Datasets:
- `syn-low-outcome-overlap` (γ_π=0, γ_ρ=4, n=10000)
- `syn-low-treatment-overlap` (γ_π=5, γ_ρ=0, n=10000)
- `syn-dual-overlap` (γ_π=5, γ_ρ=3, n=10000)
- `syn-healthy` (γ_π=0, γ_ρ=0, n=10000)

```python
import json
from pathlib import Path

OUT = Path('..') / 'outputs'
DATASETS = [
    'syn-low-outcome-overlap',
    'syn-low-treatment-overlap',
    'syn-dual-overlap',
    'syn-healthy',
]

results = {}
for ds in DATASETS:
    with open(OUT / f'{ds}_pehe_nn.json', encoding='utf-8') as f:
        results[ds] = json.load(f)

list(results.keys())
```

```python
EXCLUDE = {} #{"DO-learner-2", "TO-learner-2"}

for ds, d in results.items():
    c = d['config']
    print(f"=== {ds}  (gamma_pi={c['gamma']}, gamma_rho={c['gamma_rho']}, n={c['n']}, B={c['B']}) ===")
    rows = sorted(
        ((m, r) for m, r in d['models'].items() if m not in EXCLUDE),
        key=lambda kv: kv[1]['pehe_mean'],
    )
    print(f"{'Model':<14} {'PEHE mean':>10} {'PEHE std':>10}")
    print('-' * 40)
    for m, r in rows:
        print(f"{m:<14} {r['pehe_mean']:>10.5f} {r['pehe_std']:>10.5f}")
    print()
```
